#### 8.	Explique por que esse segundo conjunto de registros pode merecer maior atenção da equipe de energia.



A equipe de energia deve dar maior atenção a este segundo conjunto de registros (`df_consumo_pf_baixo`) porque eles representam períodos em que a **demanda de energia é alta (consumo elevado)** e, simultaneamente, o **fator de potência está baixo**. Esta combinação é particularmente problemática por várias razões:

1.  **Ineficiência Energética e Perdas:** Um baixo fator de potência indica que a energia elétrica está sendo usada de forma ineficiente. Parte da energia reativa (que não realiza trabalho útil) aumenta as perdas na rede de distribuição e nos equipamentos da indústria. Isso significa que, para o mesmo trabalho útil, a empresa está consumindo mais energia total, resultando em desperdício.
2.  **Multas e Custos Adicionais:** Empresas distribuidoras de energia frequentemente aplicam multas a consumidores que operam com fator de potência abaixo de um limite regulatório (geralmente próximo de 0.92 ou 0.95 no Brasil). Identificar e corrigir esses períodos pode evitar custos extras significativos na fatura de energia.
3.  **Sobrecarga de Equipamentos:** Um baixo fator de potência exige que a rede elétrica e os equipamentos internos (transformadores, cabos, geradores) transportem uma corrente maior para fornecer a mesma quantidade de energia ativa. Isso pode levar a superaquecimento, desgaste prematuro dos equipamentos e redução de sua vida útil, além de exigir investimentos em infraestrutura mais robusta.
4.  **Instabilidade do Sistema:** Períodos prolongados de baixo fator de potência em conjunto com alto consumo podem contribuir para a instabilidade da rede elétrica local, afetando a qualidade da energia fornecida e, em casos extremos, causando quedas de tensão ou interrupções.

Portar, esses registros são cruciais para identificar oportunidades de **otimização da qualidade da energia**, **redução de custos operacionais** e **aumento da confiabilidade e vida útil dos equipamentos** da indústria siderúrgica. A equipe pode investigar as causas específicas desses eventos (e.g., operação de motores indutivos sem compensação, uso de equipamentos antigos) e implementar soluções como bancos de capacitores para correção do fator de potência.

```python
# 1.	Renomeie as três variáveis de consumo para Consumo_Zona_1, Consumo_Zona_2 e Consumo_Zona_3.
dados = pd.read_csv('Tetuan_City_Sample.csv')
dados.rename(columns={'Zone 1 Power Consumption': 'Consumo_Zona_1'}, inplace=True)
dados.rename(columns={'Zone 2  Power Consumption': 'Consumo_Zona_2'}, inplace=True)
dados.rename(columns={'Zone 3  Power Consumption': 'Consumo_Zona_3'}, inplace=True)

dados.head()
```

Isso pode ser crucial para a equipe de planejamento de energia, pois sugere que o uso de aparelhos de refrigeração, ar condicionado ou outros equipamentos sensíveis à temperatura pode ser um fator contribuinte significativo para esses picos de consumo.

Compreender essa correlação permite à equipe focar seus esforços em estratégias de gestão da demanda que considerem as condições climáticas, como programas de eficiência energética ou incentivos para reduzir o consumo durante ondas de calor.

## Dataset 5 — Wind & Solar Energy Production (Kaggle)

fonte: https://www.kaggle.com/datasets/ahmeduzaki/wind-and-solar-energy-production-dataset

Situação
Um operador de um portfólio de fontes renováveis deseja comparar a ocorrência de períodos de alta produção solar e alta produção eólica. Como as duas fontes possuem escalas próprias, cada uma deverá ser comparada com o seu próprio valor máximo.


```python
# 1.	Renomeie as variáveis principais para Geracao_Solar e Geracao_Eolica.
dados = pd.read_csv('Wind_power_sample.csv')
dados.rename(columns={'Source': 'Recurso'}, inplace=True)
dados.rename(columns={'Production': 'Produção'}, inplace=True)

dados.head()
```

```python
# 3.	Calcule separadamente 70% do máximo solar e 70% do máximo eólico.
dmax_solar = dados[dados['Recurso'] == 'Solar']['Produção'].max()
dmax_eolico = dados[dados['Recurso'] == 'Wind']['Produção'].max()
```

```python
# 4.	Crie um DataFrame para registros de alta geração solar e outro para registros de alta geração eólica.
df70_solar = dados[(dados['Recurso'] == 'Solar') & (dados['Produção'] > dmax_solar*0.7)]
df70_eolico = dados[(dados['Recurso'] == 'Wind') & (dados['Produção'] > dmax_eolico*0.7)]
display(df70_solar.head())
display(df70_eolico.head())
```

```python
# 5.	Conte os registros de cada DataFrame e calcule seus respectivos percentuais.
num_registros_acima_limite_solar = df70_solar.shape[0]
percentual_registros_acima_limite_solar = (num_registros_acima_limite_solar / dados.shape[0]) * 100

num_registros_acima_limite_eolico = df70_eolico.shape[0]
percentual_registros_acima_limite_eolico = (num_registros_acima_limite_eolico / dados.shape[0]) * 100
```

```python
# 6.	Compare qual fonte aparece com maior frequência acima de 70% do seu próprio máximo.
if num_registros_acima_limite_solar > num_registros_acima_limite_eolico:
    print(f"A fonte 'Solar' aparece com maior frequência acima de 70% do seu próprio máximo.")
    print(f"Número de registros acima de 70% do máximo solar: {num_registros_acima_limite_solar}")
    print(f"Percentual da amostra que esses registros representam: {percentual_registros_acima_limite_solar:.2f}%")
else:
    print(f"A fonte 'Wind' aparece com maior frequência acima de 70% do seu próprio máximo.")
    print(f"Número de registros acima de 70% do máximo eólico: {num_registros_acima_limite_eolico}")
    print(f"Percentual da amostra que esses registros representam: {percentual_registros_acima_limite_eolico:.2f}%")
```

#### 7.	Explique por que não seria adequado utilizar diretamente o mesmo valor numérico de potência como limite para as duas fontes sem antes observar a escala de cada variável.

#### Explicação sobre a inadequação de um mesmo limite numérico para diferentes fontes de energia

Não seria adequado utilizar diretamente o mesmo valor numérico de potência (por exemplo, 10.000 kW) como limite para as duas fontes de energia (solar e eólica) sem antes observar a escala de cada variável por algumas razões fundamentais:

1.  **Diferenças na Capacidade Instalada e no Potencial de Geração:** A capacidade máxima de geração de uma usina solar pode ser significativamente diferente da de uma usina eólica. Cada tipo de fonte possui características intrínsecas e limites tecnológicos que determinam o seu potencial máximo de produção. Usar um limite absoluto que se baseia, por exemplo, no máximo da fonte eólica, pode fazer com que a fonte solar nunca o atinja, ou vice-versa, distorcendo a análise de 'alta geração'.

2.  **Variações na Natureza da Geração:** A geração solar depende diretamente da irradiação solar, que varia com a hora do dia, estação do ano e condições climáticas. A geração eólica depende da velocidade do vento, que também possui padrões de variabilidade distintos. Essas diferentes naturezas de geração significam que seus perfis de produção e picos são intrinsecamente diferentes.

3.  **Comparabilidade Distorcida:** Se aplicarmos um limite de 15.000 kW, por exemplo, e a usina solar tiver um máximo histórico de 12.000 kW, ela nunca atingiria o critério de 'alta geração', mesmo quando estivesse operando em sua capacidade máxima. Por outro lado, se a usina eólica tivesse um máximo histórico de 25.000 kW, 15.000 kW poderia ser considerado uma produção 'média' para ela. Isso levaria a conclusões errôneas ou comparações injustas entre as fontes.

4.  **Objetivos da Análise:** Ao comparar a 'ocorrência de períodos de alta produção', o que se busca é entender quando cada fonte está operando em um regime próximo ao seu potencial máximo ou em uma capacidade significativamente elevada *para ela mesma*. Definir um percentual do máximo individual (como 70% ou 75% do seu próprio valor máximo) permite uma análise mais justa e relevante, destacando os períodos de alta performance relativa para cada tecnologia, independentemente de suas escalas absolutas. Isso foi feito nas etapas anteriores ao calcular `dmax_solar` e `dmax_eolico` separadamente e então aplicar um percentual a cada um.

Uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio. O dataset é o mesmo utilizado como referência em aula, mas o critério de análise será diferente.


```python
# 1.	Carregue a nova amostra e organize os atributos utilizando nomes mais simples.
dados = pd.read_csv('Individual_household_sample.csv')
dados.rename(columns={'Global_active_power': 'Potencia_Ativa_Global'}, inplace=True)
dados.rename(columns={'Global_reactive_power': 'Potencia_Reactiva_Global'}, inplace=True)
dados.rename(columns={'Voltage': 'Tensão'}, inplace=True)
dados.rename(columns={'Global_intensity': 'Intensidade_Global'}, inplace=True)
dados.rename(columns={'Sub_metering_1': 'Potencia_Sub_1'}, inplace=True)
dados.rename(columns={'Sub_metering_2': 'Potencia_Sub_2'}, inplace=True)
dados.rename(columns={'Sub_metering_3': 'Potencia_Sub_3'}, inplace=True)

dados.head()
```

```python
# 2.	Determine o valor máximo de potência ativa.
dados['Potencia_Ativa_Global'].max()
```

```python
# 3.	Calcule 75% do valor máximo e crie um DataFrame contendo registros acima desse limite.
dmax = dados['Potencia_Ativa_Global'].max()
df75 = dmax*0.75
df75 = dados[dados['Potencia_Ativa_Global'] > df75]
df75.head()
```

```python
# 4.	Calcule a quantidade e o percentual de registros selecionados.
num_registros_acima_limite = df75.shape[0]
percentual_registros_acima_limite = (num_registros_acima_limite / dados.shape[0]) * 100

print(f"Número de registros com Potencia_Ativa_Global acima do limiar de 75%: {num_registros_acima_limite}")
print(f"Percentual da amostra que esses registros representam: {percentual_registros_acima_limite:.4f}")
```

```python
# 5.	Calcule a corrente média da amostra.
media_corrente = dados['Intensidade_Global'].mean()
print(f"A corrente média da amostra é: {media_corrente:.2f}")
```

```python
# 6.  Crie um segundo DataFrame contendo simultaneamente potência ativa acima de 75% do máximo e corrente acima da média.
limiar_potencia_ativa_alta = dados['Potencia_Ativa_Global'].max() * 0.75
df_consumo_corrente_acima_media = dados[(dados['Potencia_Ativa_Global'] > limiar_potencia_ativa_alta) & (dados['Intensidade_Global'] > media_corrente)]
display(df_consumo_corrente_acima_media.head())
```

#### 7.	Compare os dois conjuntos e explique o efeito da inclusão da corrente como segunda condição.

```python
num_registros_duplo_criterio_corrente = df_consumo_corrente_acima_media.shape[0]
percentual_duplo_criterio_corrente = (num_registros_duplo_criterio_corrente / dados.shape[0]) * 100

print(f"Número de registros com Potencia_Ativa_Global acima do limiar de 75% E Intensidade_Global acima da média: {num_registros_duplo_criterio_corrente}")
print(f"Percentual da amostra que esses registros representam (com duplo critério): {percentual_duplo_criterio_corrente:.4f}%")
```

Comparando os dois conjuntos de registros:

*   **`df75` (apenas consumo ativo alto):** Contém **58 registros**, representando **0.0283%** da amostra total.
*   **`df_consumo_corrente_acima_media` (consumo ativo alto E corrente acima da média):** Contém **58 registros**, representando **0.0283%** da amostra total.

### Efeito da Inclusão da Condição Ambiental (Corrente)

Neste caso específico, a inclusão da condição de `Intensidade_Global` acima da média **não alterou o número de registros selecionados** em relação ao filtro apenas por `Potencia_Ativa_Global` acima de 75% do máximo. Isso sugere que, para este dataset e para o critério de alta potência ativa definido:

1.  **Correlação Forte:** Os eventos de `Potencia_Ativa_Global` muito alta estão altamente correlacionados com `Intensidade_Global` acima da média. Ou seja, sempre que a potência ativa excede 75% do seu máximo, a corrente também já está acima da média.
2.  **Redundância Parcial do Critério:** A segunda condição (`Intensidade_Global` > `media_corrente`) não está excluindo registros adicionais porque o conjunto de registros que já atende ao primeiro critério (`Potencia_Ativa_Global` > 75% do máximo) já está, em sua totalidade, atendendo também ao segundo critério.

Para a equipe de monitoramento, isso pode indicar que a `Intensidade_Global` acima da média não atua como um filtro adicional para identificar um subconjunto mais específico de eventos de interesse quando a potência ativa já está em níveis muito elevados. Em outras palavras, um *alto consumo de eletrodomésticos* (potência ativa) já é um bom indicativo de *alta corrente* neste cenário.
